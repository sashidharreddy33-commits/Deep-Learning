```python
import numpy as np
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.optimizers import Adam
data = np.random.rand(1000, 50)
data = data / np.max(data)
input_dim = 50
encoding_dim = 16
input_layer = Input(shape=(input_dim,))
encoded = Dense(32, activation='relu')(input_layer)
encoded = Dense(encoding_dim, activation='relu')(encoded)
decoded = Dense(32, activation='relu')(encoded)
decoded = Dense(input_dim, activation='sigmoid')(decoded)
autoencoder = Model(inputs=input_layer, outputs=decoded)
autoencoder.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
autoencoder.fit(data, data,
                epochs=20,
                batch_size=32,
                shuffle=True)
encoder = Model(inputs=input_layer, outputs=encoded)
latent_sample = np.random.rand(1, encoding_dim)
decoder_input = Input(shape=(encoding_dim,))
decoder_layer1 = autoencoder.layers[-2](decoder_input)
decoder_output = autoencoder.layers[-1](decoder_layer1)
decoder = Model(decoder_input, decoder_output)
generated_music = decoder.predict(latent_sample)
print("Generated Music Sequence:")
print(generated_music)
```
